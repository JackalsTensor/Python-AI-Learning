温度数据分析

```python
import pandas as pd
import numpy as np
temperatures=pd.Series([28,31,29,32,30,27,33],index=['周一','周二','周三','周四','周五','周六','周日'])
#找到温度超过30度的天数
n=(temperatures[temperatures>30].count())
print(f"超过30天的天数为{n}天")
print("\n")
#将温度从高到低排序
t2=temperatures.sort_values(ascending=False)
print(f"从高到低的排序为\n{t2}")
print("\n")
#找到温度变化最大的两天
t3=temperatures.diff().abs()
t4=t3.sort_values(ascending=False).keys()
t5=t4[:2].tolist()
print(f"温度变化最大的两天为{t5}")
```

股票价格分析

```python
prices=pd.Series([102.3,103.5,105.1,104.8,106.2,107.0,106.5,108.1,109.3,110.2],index=pd.date_range('2023-01-01',periods=10))
print(prices)
```

```python
#计算每日收益率
a=prices.pct_change()
print(a)
```

```python
#收益率最高的日期
print(a.idxmax())
#收益率最低的日期
print(a.idxmin())
#波动率(收益率的标准差)
print(a.std())
```

```python
sales=pd.Series([120,135,145,160,155,170,180,175,190,200,210,220],index=pd.date_range('2022-01-01',periods=12,freq='MS'))
print(sales)
```

```python
#季度的平均销量
sales.resample('QS').mean()
```

```python
print('销量最高的月份',sales.idxmax())
```

```python
print('月环比的增长率')
sales.pct_change()
```

```python
#连续增长超过两个月的月份
a=sales.pct_change()
b=a>0
b[b.rolling(3).sum()==3].keys().tolist()
```

每小时销售数据分析

```python
import pandas as pd
import numpy as np
np.random.seed(42)
hours_sales=pd.Series(np.random.randint(0,100,24),index=pd.date_range('2025-01-01',periods=24,freq='h'))
#每日总销售额
day_sales=hours_sales.resample('D').sum()
#计算每天营业时间和非营业时间销售额比例
hours_sales.between_time('8:00','22:00')#筛选一段时间内的series
business_hours_sales=hours_sales[(hours_sales.index.hour>=8)&((hours_sales.index.hour<=22))].sum()
print(business_hours_sales/(day_sales-business_hours_sales))
```

```python
#绘制折线图
import matplotlib.pyplot as plt
from matplotlib import rcParams #设置中文字体
rcParams['font.sans-serif'] = ['SimHei']
#创建图表设置大小
plt.figure(figsize=(10,5))
month=['1月','2月','3月','4月']
sales=[100,150,80,130]
plt.plot(month,sales,label='产品A')
#添加标题
plt.title('2025年销量趋势',color='red',fontsize=20)
#添加坐标轴标签
plt.xlabel('月份',fontsize=10,color='red')
plt.ylabel('销售额（万元）',fontsize=10,color='red')
#添加图例
plt.legend(loc='upper left')
#添加网图线
plt.grid(True,alpha=0.5,color='black',linestyle='--')
#设置刻度字体大小
plt.xticks(rotation=0,fontsize=10)
plt.yticks(rotation=0,fontsize=10)
#设置y轴的范围
plt.ylim(50,180)
#在每个数据点上显示数值
for x,y in zip(month,sales):
    print(x,y)
    plt.text(x,y,str(y),ha='center',va='bottom',fontsize=10)
#显示图标
plt.show()
```

```python
#柱状图模板
import matplotlib.pyplot as plt
from matplotlib import rcParams #设置中文字体
rcParams['font.sans-serif'] = ['SimHei']
#创建图表设置大小
plt.figure(figsize=(10,5))
#要绘制的数据
subjects=['语文','数学','英语','科学']
scores=[85,92,78,88]
#绘制柱状图
plt.bar(subjects,scores,label='小红')
#添加标题
plt.title('小红2025年成绩分布',color='red',fontsize=20)
#添加坐标轴的标签
plt.xlabel('科目',fontsize=10)
plt.ylabel('分数',fontsize=10)
#添加图例
plt.legend(loc='upper right')
#添加网格线
plt.grid(axis='y',alpha=0.5,color='blue',linestyle='--')
#设置y轴的范围
plt.ylim(0,100)
#在每个数据点上设置数值
for x,y in zip(subjects,scores):
    plt.text(x,y,str(y),ha='center',va='bottom',fontsize=10)
#自动优化排版
plt.tight_layout()
#显示图标
plt.show
```

```python
#条形图模板
import matplotlib.pyplot as plt
from matplotlib import rcParams #设置中文字体
rcParams['font.sans-serif'] = ['SimHei']
#创建图表设置大小
plt.figure(figsize=(10,5))
country=['United States','China','Japan','Germany','India']
GDP=[85,92,78,88,60]
#添加标题
plt.title('2025年GDP排名',color='red',fontsize=20)
#添加坐标轴的标签
plt.xlabel('GDP',fontsize=10)
plt.ylabel('国家',fontsize=10)
plt.tight_layout()
#绘制条形图
plt.barh(country,GDP)
plt.show()
```

```python
#绘制饼图
import matplotlib.pyplot as plt
from matplotlib import rcParams #设置中文字体
rcParams['font.sans-serif'] = ['SimHei']
#创建图表设置大小
plt.figure(figsize=(10,5))
#要绘制的数据
things=['学习','娱乐','运动','睡觉','其他']
times=[6,4,1,8,5]
colors=['red','green','blue','yellow','pink']
plt.pie(times,labels=things,autopct='%1.1f%%',startangle=90)
#添加标题
plt.title('一天时间分布',color='red',fontsize=20)
#自动优化排版
plt.tight_layout()
plt.show()
```

```python
#绘制散点图
import matplotlib.pyplot as plt
from matplotlib import rcParams
import random
rcParams['font.sans-serif'] = ['SimHei']
plt.figure(figsize=(10,5))
x=[]
y=[]
for i in range(1000):
    tmp=random.uniform(0,10)
    x.append(tmp)
    tmp2=2*tmp+random.gauss(0,2)
    y.append(tmp2)
plt.scatter(x,y,color='blue',alpha=0.5,s=30,label='数据')
plt.title('x变量与y变量的关系',color='red',fontsize=20)
plt.xlabel('x',fontsize=10)
plt.ylabel('y',fontsize=10)
plt.grid(True,alpha=0.5,color='black',linestyle='--')
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.plot([0,10],[0,20]) #回归线
plt.ylim(0,30)
plt.show()
```

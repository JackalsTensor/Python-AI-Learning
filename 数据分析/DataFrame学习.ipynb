```python
import pandas as pd
import numpy as np
s1=pd.Series([1,2,3,4,5])
s2=pd.Series([6,7,8,9,10])
# df=pd.DataFrame({"第一列":s1,"第二列":s2})
df=pd.DataFrame(
    {
        "id":[1,2,3,4,5],
        "name":["潘思雨","朱亮星","FKM","大斌","老pie"],
        "age":[18,19,18,26,50],
        "weight(kg)":[70,150,70,180,90],
        "score":[514,503,516,524,505]
    },index=[1,2,3,4,5],columns=["name","weight(kg)","age","score"]
)
print(df)
print('行索引:')
print(df.index)
print('列标签:')
print(df.columns)
print('值：')
print(df.values)
print('维度：',df.ndim)
print('形状:',df.shape)
print('元素个数:',df.size)
print(df.dtypes)
print(df.T)
print(df.head(1))
print(df['score'].sum())
print(df.score.min())
print(df.score.median())
print(df.score.std())
print(df.describe())
print(df.count())#每一列非缺失值个数
print(df.sample(2))
print(df.cumsum())
print(df.cummin())
print(df.sort_values(by='score'))
```

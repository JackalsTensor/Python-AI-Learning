# 第一题. NumPy 数组基础操作（Series 底层依赖)


### 给定列表 data = [12, 34, 56, 78, 90, 23, 45, 67]，完成以下操作：
 1. 将列表转换为 NumPy 数组 arr，并查看数组的形状（shape）、数据类型（dtype）；
2. 计算数组的平均值、中位数、标准差；
3. 筛选出数组中 大于 50 且小于 80 的元素;
4. 将数组按 降序排序 后，取前 3 个元素组成新数组。

```python
import numpy as np
import pandas as pd
s=pd.Series([12,34,56,78,90,23,45,67])
arr=np.array(s)
#-------------------------------------------
#第一题
print('data=',arr)
print(arr.dtype)
print(arr.shape)
#--------------------------------------------
#第二题
print(arr.sum())
print(arr.mean())
print(arr.std())
#---------------------------------------------
#第三题
result=arr[(arr>50)&(arr<80)]
print(result)
#---------------------------------------------
#第四题
sorted_arr=np.sort(arr)[::-1]
print(sorted_arr[:3])
```

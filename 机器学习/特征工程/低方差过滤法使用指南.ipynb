```python
import pandas as pd
from sklearn.feature_selection import VarianceThreshold
```

```python
data={
    'feat1':[1,1,1,1],
    'feat2':[1,2,3,4],
    'feat3':[2,2,3,3],
    'feat4':[5,5,5,6]
}
df=pd.DataFrame(data)
print(df)
```

```python
vt=VarianceThreshold(0)
df_filtered=vt.fit_transform(df)
print(df_filtered)
```

```python
keep_cols=df.columns[vt.get_support()]
print(keep_cols)
```

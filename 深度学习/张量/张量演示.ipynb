```python
#演示创建张量
import torch
import numpy as np
t1=torch.tensor(10)
print(t1)
print(t1.dtype)
data=[[1,2,3],[4,5,6]]
t2=torch.tensor(data)
print(t2)
print(t2.dtype)
data=np.random.randint(0,10,(3,4))
t3=torch.tensor(data)
print(t3)
print(t3.dtype)
```

```python
#演示创建张量
import torch
import numpy as np
t1=torch.Tensor(10)
print(t1)
print(t1.dtype)
data=[[1,2,3],[4,5,6]]
t2=torch.Tensor(data)
print(t2)
print(t2.dtype)
data=np.random.randint(0,10,(3,4))
t3=torch.Tensor(data)
print(t3)
print(t3.dtype)
t4=torch.Tensor(2,3)
print(t4)
print(t4.dtype)
```

```python
#演示创建张量
import torch
import numpy as np
t1=torch.IntTensor(10)
print(t1)
print(t1.dtype)
data=[[1,2,3],[4,5,6]]
t2=torch.IntTensor(data)
print(t2)
print(t2.dtype)
data=np.random.randint(0,10,(3,4))
t3=torch.IntTensor(data)
print(t3)
print(t3.dtype)
t4=torch.FloatTensor(2,3)
print(t4)
print(t4.dtype)
```
